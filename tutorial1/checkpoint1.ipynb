Let's compare Michigan to Ohio. Which state has the higher average across all years in the test you chose?

```python
# @@@ 2
avgOhio = df[df["STATE"] == "OHIO"]["AVG_MATH_8_SCORE"].mean()
avgMichigan = df[df["STATE"] == "MICHIGAN"]["AVG_MATH_8_SCORE"].mean()

if avgOhio > avgMichigan:
    print("Ohio")
elif avgOhio < avgMichigan:
    print("Michigan")
```

<h2> Feature Engineering </h2>

After exploring the data, you can choose to modify features that you would use to predict the performance of the students on your chosen response variable.

```python
# @@@ 5
df["EXPENDITURE_PER_STUDENT"] = df["TOTAL_EXPENDITURE"] / df["ENROLL"]
df[["STATE", "YEAR", "EXPENDITURE_PER_STUDENT", "AVG_MATH_8_SCORE"]].head()
```

Feature engineering justification: Different states in the dataset have different numbers of students. I made a expenditure per student feature in order to compare educational spending relative to the number of students. This allows me to analyze its relationship with the average math 8 scores.

**Visualization 1**

```python
# @@@ 6
import matplotlib.pyplot as plt

plt.scatter(df["EXPENDITURE_PER_STUDENT"], df["AVG_MATH_8_SCORE"])
plt.xlabel("Expenditure per Student")
plt.ylabel("Average Math 8 Score")
plt.title("Expenditure/student vs Academic performance")
plt.show()
```
